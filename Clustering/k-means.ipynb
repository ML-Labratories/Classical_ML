**Image Compression**

```python
import cv2
import numpy as np
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import structural_similarity as ssim
```

```python
image = cv2.imread('img/flower_1.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
# Reshape the image to a 2D array of pixels and 3 color values (RGB)
pixels = image.reshape(-1, 3)
print(pixels.shape)
```

```python
# Normalize pixel values
pixels = pixels / 255.0
```

```python
k = 16  # Number of colors
kmeans = KMeans(n_clusters=k, random_state=42)
kmeans.fit(pixels)
```

```python
# Replace each pixel with the centroid of its cluster
compressed_pixels = kmeans.cluster_centers_[kmeans.labels_]
compressed_pixels = np.clip(compressed_pixels * 255, 0, 255)  # Rescale to 0-255
compressed_image = compressed_pixels.reshape(image.shape).astype('uint8')
```

```python
# Calculate PSNR and SSIM
original_image_gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
compressed_image_gray = cv2.cvtColor(compressed_image, cv2.COLOR_RGB2GRAY)
psnr_value = psnr(original_image_gray, compressed_image_gray, data_range=255)
ssim_value = ssim(original_image_gray, compressed_image_gray, data_range=255)
# Display the results
print(f"PSNR: {psnr_value:.2f} dB")
print(f"SSIM: {ssim_value:.4f}")
```

```python
# Display the original and compressed images
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(image)
plt.axis('off')
plt.subplot(1, 2, 2)
plt.title(f"Compressed Image with {k} colors")
plt.imshow(compressed_image)
plt.axis('off')
plt.show()
```
